# Genealogy OCR Pipeline — Scheduled (Nightly Batch)

Transcribes scanned genealogy documents from Google Drive using Vertex AI Gemini.
Designed to run as a nightly Databricks Job across all unprocessed files (limited to a specified maximum).

**Parameters:**
- `max_files`: Maximum number of files to process per run (default 50).
  Candidates are drawn in order of folder then filename across all listed folders,
  excluding any file already logged as success. Adjust to control nightly cost.

**Source table:** `workspace.staging_google_drive.documents`
**Output tables:** `workspace.genealogy.ocr_transcriptions`, `workspace.genealogy.ocr_processing_log`

**Change log:**
- v1–v3.4: See original ocr_pipeline.py
- v3.5: Scheduled batch variant. folder_name widget replaced by max_files widget (int, default 50).
        Candidates are collected across all listed folders, sorted by folder/filename,
        then sliced to max_files before processing begins. Token usage summary
        updated to cover all processed folders in this run.
- v4: Person-mention extraction folded into the same Gemini call, at transcription
      time, instead of a later manual Claude backfill pass (see Document Structuring
      — Extraction Conventions, Asana 1218421366058506). Every page now gets a
      `silver_transcript_person_mention` write (possibly zero rows) alongside its
      `ocr_transcriptions` row, so there is no longer a gap for a genuinely empty
      or unreviewed page to fall into. New `content_flag`/`content_note` mark a
      page with no extractable content or deliberately excluded; `review_status`/
      `reviewed_at`/`review_notes` track manual review (unified with the existing
      `confidence_flag` signal); `last_matched_at` lets the matching notebook
      detect a page reviewed since it was last processed. `DOC_TYPE_VOCABULARY`
      extended (TitheApportionment, AsylumRecord, PropertyConveyance,
      PersonalCorrespondence); new `ROLE_GUIDANCE` table drives a two-step prompt
      (classify doc_type_detected, then apply that type's role_in_record
      vocabulary/scoping) instead of one flat rule set for every document type.
- v4.1: `FAMILY_RELATION_ROLES` shared vocabulary + loosened role_in_record
      instructions (see ROLE_GUIDANCE cell), reconciled against the real
      distinct (doc_type_detected, role_in_record) pairs from the full
      backfill. Fixed a bug where MarriageCertificate was wrongly told not to
      record Wife/Husband mentions at all (that exclusion belongs to
      notebook_01's Cell 5e matching stage, not extraction).
- v4.2: Bounded retry for repeatedly-failing files. New `max_retries` widget
      (default 3). A file that keeps hitting the same transient error (e.g.
      Gemini 504 timeouts on an oversized PDF) now gets a distinct
      `failed_permanent` status once it's failed MAX_RETRIES times, instead
      of retrying forever or needing a manual status edit to stop it —
      `failed_permanent` is excluded from future candidate runs same as
      `success`/`skipped`, but stays visually and structurally distinct from
      both (not a content decision like `skipped`, not still-retryable like
      `error`). Attempt counts are read from `ocr_processing_log`'s existing
      per-attempt rows, no new column needed.

````python
# Magic byte signatures → (mime_type, canonical_extension)
# GIFs are detected but then converted to JPEG before sending to Gemini —
# image/gif support is inconsistent across Gemini model versions (can return 400).
MAGIC_SIGNATURES = [
    (b'\xff\xd8\xff',         "image/jpeg",       ".jpg"),
    (b'\x89PNG\r\n\x1a\n',    "image/png",        ".png"),
    (b'II*\x00',              "image/tiff",       ".tif"),   # TIFF little-endian
    (b'MM\x00*',              "image/tiff",       ".tif"),   # TIFF big-endian
    (b'%PDF',                 "application/pdf",  ".pdf"),
    #(b'GIF87a',               "image/gif",        ".gif"),
    #(b'GIF89a',               "image/gif",        ".gif"),
]

EXTENSION_MIME_MAP = {
    ".jpg":  "image/jpeg",
    ".jpeg": "image/jpeg",
    ".png":  "image/png",
    ".tif":  "image/tiff",
    ".tiff": "image/tiff",
    ".pdf":  "application/pdf",
    #".gif":  "image/gif",
}


def get_mime_type(file_path: str):
    """
    Determine MIME type and canonical extension for a file.
    Uses the file extension if recognised; falls back to magic byte detection.
    Returns (mime_type, ext) or None if the file is unsupported.
    """
    ext = Path(file_path).suffix.lower()
    if ext in EXTENSION_MIME_MAP:
        return EXTENSION_MIME_MAP[ext], ext

    # No recognised extension — inspect the first 8 bytes
    try:
        with open(file_path, "rb") as f:
            header = f.read(8)
        for magic, mime, canon_ext in MAGIC_SIGNATURES:
            if header[:len(magic)] == magic:
                return mime, canon_ext
    except Exception:
        pass

    return None
````


Selects all files from the documents table not yet logged as success,
sorted by path for deterministic ordering, sliced to MAX_FILES.
The job automatically continues where it left off on each subsequent run.


````python
source_df = spark.sql(f"""
    SELECT
        file_id,
        '{VOLUME_BASE}/' || _fivetran_file_path AS file_path,
        _fivetran_file_path AS relative_path,
        size,
        modified_at
    FROM {SOURCE_TABLE}
    WHERE _fivetran_deleted = false
""")

# 'failed_permanent' (bounded-retry giveup, see the main loop's exception
# handler below) is excluded alongside success/skipped -- a file that has
# already failed MAX_RETRIES times stops resurfacing as a candidate. A plain
# 'error' status (fewer than MAX_RETRIES attempts so far) is NOT excluded,
# so it keeps being retried on subsequent runs until it succeeds or gives up.
processed_ids = spark.sql(f"""
    SELECT file_id FROM {PROC_LOG_TBL}
    WHERE status in ('success', 'skipped', 'failed_permanent')
""")

to_process_df = (
    source_df
    .join(processed_ids, on="file_id", how="left_anti")
    .orderBy("relative_path")
    .collect()
)[:MAX_FILES]

total = len(to_process_df)
print(f"Candidate files (capped at {MAX_FILES}): {total}")
for row in to_process_df:
    print(f"  {row.file_path}  ({row.size:,} bytes)")

# Prior 'error' attempt counts per file_id, computed once up front rather
# than re-queried on every failure inside the loop. Incremented in memory as
# new errors occur this run, so repeated failures of the same file within
# one run are still counted correctly toward MAX_RETRIES.
prior_error_counts_df = spark.sql(f"""
    SELECT file_id, COUNT(*) AS n
    FROM {PROC_LOG_TBL}
    WHERE status = 'error'
    GROUP BY file_id
""")
error_attempt_counts = {row.file_id: row.n for row in prior_error_counts_df.collect()}
````

````python
# Document type keyword → (specialist description, era hint, key fields)
DOC_TYPE_MAP = {
    "census": (
        "census enumerator's return",
        "19th or early 20th century",
        "name, age, sex, occupation, birthplace, relationship to head of household, marital status"
    ),
    "birthcertificate": (
        "birth certificate", "19th or early 20th century",
        "child's name, date and place of birth, father's name and occupation, mother's name and maiden name, informant"
    ),
    "birth": (
        "birth certificate", "19th or early 20th century",
        "child's name, date and place of birth, father's name and occupation, mother's name and maiden name, informant"
    ),
    "marriagecertificate": (
        "marriage certificate", "19th or early 20th century",
        "names of bride and groom, ages, occupations, fathers' names and occupations, witnesses, officiant, date and place"
    ),
    "marriage": (
        "marriage certificate", "19th or early 20th century",
        "names of bride and groom, ages, occupations, fathers' names and occupations, witnesses, officiant, date and place"
    ),
    "deathcertificate": (
        "death certificate", "19th or early 20th century",
        "name of deceased, date and place of death, age, occupation, cause of death, informant"
    ),
    "death": (
        "death certificate", "19th or early 20th century",
        "name of deceased, date and place of death, age, occupation, cause of death, informant"
    ),
    "will": (
        "last will and testament", "19th or early 20th century",
        "testator's name, beneficiaries, executors, bequests, date of signing, witnesses"
    ),
    "probate": (
        "probate record", "19th or early 20th century",
        "deceased's name, estate value, executors, date of probate, court"
    ),
    "grant": (
        "grant of probate", "19th or early 20th century",
        "deceased's name, estate value, executors, date of grant, court"
    ),
    "newspaper":  ("newspaper clipping", "Victorian or Edwardian era", "all text as printed including headline and captions"),
    "gazette":    ("newspaper or gazette clipping", "Victorian or Edwardian era", "all text as printed"),
    "advertiser": ("newspaper clipping", "Victorian or Edwardian era", "all text as printed"),
    "herald":     ("newspaper clipping", "Victorian or Edwardian era", "all text as printed"),
    "observer":   ("newspaper clipping", "Victorian or Edwardian era", "all text as printed"),
    "times":      ("newspaper clipping", "Victorian or Edwardian era", "all text as printed"),
    "baptism": (
        "baptism register entry", "18th or 19th century",
        "child's name, date of baptism, parents' names, father's occupation, parish"
    ),
    "bapt": (
        "baptism register entry", "18th or 19th century",
        "child's name, date of baptism, parents' names, parish"
    ),
    "burial": (
        "burial register entry", "18th or 19th century",
        "name of deceased, date of burial, age, abode, officiant, parish"
    ),
    "electoral": (
        "electoral register entry", "19th or early 20th century",
        "name, address, qualification, nature of qualification"
    ),
    "military": (
        "military service record", "World War I or World War II era",
        "name, rank, regiment, service number, date of enlistment, postings, medical notes"
    ),
    "poorlaw": (
        "Poor Law parochial relief application", "19th century",
        "applicant name, family members, ages, address, parish, reason for application, decision"
    ),
    "poor": (
        "Poor Law parochial relief application", "19th century",
        "applicant name, family members, ages, address, parish, reason for application, decision"
    ),
    "bmd": (
        "civil registration BMD index entry", "19th or early 20th century",
        "name, registration district, volume, page number, year and quarter of registration"
    ),
    "gro": (
        "civil registration BMD index entry", "19th or early 20th century",
        "name, registration district, volume, page number, year and quarter of registration"
    ),
}


def parse_filename(filename: str) -> dict:
    """
    Parse a genealogy document filename into structured metadata.
    Convention: SURNAME_Firstname_Year_Description.ext
    Year may be YYYY or YYYY-MM-DD.
    """
    stem = Path(filename).stem
    parts = stem.split("_")

    result = {"surname": None, "forename": None, "year": None, "doc_type": None, "parse_ok": False}

    if len(parts) < 3:
        result["doc_type"] = stem
        return result

    result["surname"]  = parts[0]
    result["forename"] = parts[1]

    year_token = parts[2]
    if re.match(r'^\d{4}(-\d{2}(-\d{2})?)?$', year_token):
        result["year"]     = year_token
        result["doc_type"] = "_".join(parts[3:]) if len(parts) > 3 else "Unknown"
    else:
        result["doc_type"] = "_".join(parts[2:])

    result["parse_ok"] = bool(result["surname"] and result["forename"] and result["year"])
    return result


def get_doc_type_context(doc_type: str) -> tuple:
    """Map a raw doc_type token to (description, era, fields). Falls back to generic."""
    if not doc_type:
        return ("historical document", "19th or early 20th century", "all visible text and fields")
    key = re.sub(r'[^a-z]', '', doc_type.lower())
    for pattern, context in DOC_TYPE_MAP.items():
        if pattern in key:
            return context
    return ("historical document", "19th or early 20th century", "all visible text and fields")
````

````python
# role_in_record guidance keyed by doc_type_detected (Gemini's own classification
# from this same call — distinct from DOC_TYPE_MAP above, which is keyed by the
# raw filename-parsed doc_type token and only builds transcription context).
#
# Two layers, not one flat per-type list — v4.1 revision after Ed pulled the
# actual distinct (doc_type_detected, role_in_record) pairs that came up across
# the full Sept 2026 Claude backfill (2,807 rows). That data showed the same
# ~30 blood/in-law/household relation terms recurring across nearly every doc
# type (a DeathCertificate's informant can be a Son-in-law; a NewspaperClipping
# death notice lists survivors; a MonumentalInscription names a Wife) — so
# family relations are one shared vocabulary (FAMILY_RELATION_ROLES) available
# for every type, and each doc_type_detected entry below only lists its
# ADDITIONAL non-family roles (Deceased, Witness, Executor, Elector, ...) plus
# any scoping/notes specific to that type. Consolidated from Document
# Structuring — Extraction Conventions (Asana 1218421366058506) and the real
# backfill data. Extend this (and the source doc) when a genuinely new case
# needs a role that isn't here yet.
FAMILY_RELATION_ROLES = [
    "Head", "Wife", "Husband", "Son", "Daughter", "Father", "Mother", "Child",
    "Brother", "Sister", "Grandparent", "Grandchild", "Stepchild",
    "Aunt", "Niece", "Nephew", "Cousin",
    "Father in law", "Mother in law", "Son in law", "Daughter in law",
    "Brother in law", "Sister in law", "Child in law",
    "Servant", "Lodger", "Boarder", "Visitor", "Resident",
]

ROLE_GUIDANCE = {
    "Census": {
        "extra_roles": [],
        "scoping": "Only the document's own household/family — never neighbouring households on the same page.",
        "notes": "If there is no explicit relation-to-head column (e.g. 1841 Census, 1939 Register), infer role from schedule order, marital status, and age pattern; mark confidence medium at best and say in notes that the role was inferred. Never force a role that contradicts age plausibility.",
    },
    "BirthCertificate": {
        "extra_roles": [],
        "scoping": "The informant's stated relation to the child, if given (often Grandparent), beyond the child/parents already covered by the primary document match.",
    },
    "MarriageCertificate": {
        "extra_roles": ["Witness"],
        "scoping": "Bride, groom (Wife/Husband), both fathers, witnesses — record all of these normally. (The primary spouse match itself is resolved more reliably elsewhere downstream with no OCR dependency, but that is a later-stage concern; extraction should still capture what this document actually states.)",
    },
    "DeathCertificate": {
        "extra_roles": ["Deceased"],
    },
    "BaptismRegister": {
        "extra_roles": [],
        "notes": "If a parent's surname (commonly the mother's) is not stated, infer it from another named family member in the same record (usually the father); record the inferred surname in name_raw and flag the inference in notes.",
    },
    "BurialRegister": {
        "extra_roles": ["Deceased"],
    },
    "Will": {
        "extra_roles": ["Testator", "Executor", "Legatee", "Witness", "Deceased"],
        "notes": "The Testator is this document's own primary subject, already identified from the filename/context above — just tag their role as Testator, do not treat them as a separate person to disambiguate.",
    },
    "Probate": {
        "extra_roles": ["Executor", "Witness", "Testator", "Deceased"],
    },
    "NewspaperClipping": {
        "extra_roles": ["Deceased", "Witness", "Defendant", "Complainant", "Soldier", "Occupier"],
        "scoping": "Only the relevant entry/notice — not the whole column of unrelated notices on the same clipping.",
    },
    "ElectoralRegister": {
        "extra_roles": ["Elector"],
    },
    "MilitaryRecord": {
        "extra_roles": ["Soldier"],
    },
    "BMDIndex": {
        "extra_roles": ["Deceased"],
    },
    "PoorLawRecord": {
        "extra_roles": ["Applicant"],
    },
    "MonumentalInscription": {
        "extra_roles": ["Deceased"],
    },
    "PassengerList": {
        "extra_roles": ["Passenger"],
    },
    "Directory": {
        "extra_roles": [],
        "scoping": "Only the subject's own line(s) in a large list document (directory, electoral roll, poll book, pay list) — not the hundreds of other entries on the page.",
    },
    "CV": {
        "extra_roles": [],
    },
    "CriminalRecord": {
        "extra_roles": ["Defendant", "Complainant"],
    },
    "TitheApportionment": {
        "extra_roles": ["Landowner", "Occupier"],
    },
    "AsylumRecord": {
        "extra_roles": ["Patient"],
    },
    "PropertyConveyance": {
        "extra_roles": ["Grantor", "Grantee", "Trustee"],
    },
    "PersonalCorrespondence": {
        "extra_roles": ["Recipient"],
    },
    "Other": {
        "extra_roles": ["Declarant", "Witness", "Testator", "Deceased"],
        "notes": "Free-form legal/institutional documents not covered by a more specific type (e.g. a name-change deed or self-declaration) tend to land here — apply the closest-fitting roles above plus family relations, and use judgement for anything that still doesn't fit.",
    },
}


def build_role_guidance_block() -> str:
    """
    Render ROLE_GUIDANCE into a reference block for the prompt — the shared
    family-relation vocabulary once, then every doc_type_detected value's
    additional roles/scoping/notes, so Gemini can look up its own
    classification's rules after choosing doc_type_detected (see build_prompt,
    "Person Extraction — do this in two steps").
    """
    lines = [
        "Family/household relation roles (available for EVERY doc_type_detected, not just types listed below):",
        "  " + ", ".join(FAMILY_RELATION_ROLES),
        "",
        "Additional roles and rules specific to each doc_type_detected:",
    ]
    for doc_type in DOC_TYPE_VOCABULARY:
        g = ROLE_GUIDANCE.get(doc_type, {"extra_roles": []})
        extras = ", ".join(g["extra_roles"]) if g["extra_roles"] else "none — family relations above cover this type"
        lines.append(f"- {doc_type}: additional roles = {extras}")
        if g.get("scoping"):
            lines.append(f"    scoping: {g['scoping']}")
        if g.get("notes"):
            lines.append(f"    notes: {g['notes']}")
    return "\n".join(lines)
````

````python
def build_prompt(metadata: dict) -> str:
    """
    Build a Gemini transcription + person-extraction prompt from parsed filename metadata.

    Key prompt design principles (v3):
    - JSON-only instruction is both positive ("return only") AND negative ("do not wrap",
      "do not add commentary") to reduce fence/preamble failures
    - Output skeleton appears at the very end of the prompt — LLMs weight end-of-prompt heavily
    - Vocabulary list appears immediately before the skeleton so doc_type_detected
      values are fresh in context when the skeleton is read

    v4 — person-mention extraction folded into the same call (previously a separate
    manual Claude backfill pass; see Document Structuring — Extraction Conventions,
    Asana 1218421366058506). Two-step instruction: classify doc_type_detected first,
    then look up and apply that type's rules from the Person-Extraction Guide below —
    a single-pass Gemini call self-selects the relevant guidance rather than the
    pipeline branching the prompt text per type.
    """
    doc_desc, era, fields = get_doc_type_context(metadata.get("doc_type"))

    subject_hint = ""
    if metadata.get("forename") and metadata.get("surname"):
        subject_hint = f"The document relates to {metadata['forename']} {metadata['surname']}"
        if metadata.get("year"):
            subject_hint += f", circa {metadata['year']}"
        subject_hint += "."

    vocab_list = "\n".join(f"  - {v}" for v in DOC_TYPE_VOCABULARY)
    role_guidance_block = build_role_guidance_block()

    return f"""You are an expert historical document transcriptionist specialising in {era} {doc_desc}s and the handwriting styles of that period.

{subject_hint}

## Transcription Rules

- Preserve original spelling exactly, including phonetic or archaic spellings (e.g. "labrer", "Willm", "Thos")
- Mark any word you cannot read with confidence as [?]
- For partially legible words, include your best reading followed by [?] (e.g. "Thos[?]")
- Do not modernise, correct, or standardise any spelling or punctuation
- Preserve abbreviations as written
- Transcribe all fields visible in the document, paying particular attention to: {fields}
- For tabular or multi-column documents, preserve the row and column structure in your transcription

## Person Extraction — do this in two steps

Step 1: Classify this page using doc_type_detected (vocabulary given below).

Step 2: Find your chosen doc_type_detected in the Person-Extraction Guide below.
Every role_in_record you assign should normally come from one of two places: the
shared family/household relation list (available for every doc type — Wife,
Son-in-law, Grandchild, Lodger, and so on), or that specific doc_type_detected's
own additional roles (e.g. Executor/Witness for a Will, Elector for an Electoral
Register). If neither list has a good fit for a person's actual stated role,
use your own concise, reasonable term rather than forcing a bad match — but
stay close in spirit to the vocabulary already established (prefer an existing
near-equivalent, e.g. "Cousin", over inventing something idiosyncratic).

### Core principles (apply regardless of doc_type_detected)

- One entry per person actually named on THIS page for the document's own
  relevant subject — not neighbouring households, unrelated newspaper
  notices on the same clipping, or the hundreds of other entries on a
  directory/electoral/pay-list page.
- name_raw: forename-first order, regardless of the source document's own
  column order (e.g. a "Surname, Forename" register becomes "Forename Surname").
- Preserve spelling exactly as transcribed — do NOT silently correct a
  surname or forename to a "more likely" spelling. A misread or archaic
  spelling gets fixed later via a person alias, not here.
- Ditto marks ("Do", "do") resolve to the last full name/surname given
  above in the same document — record the RESOLVED name in name_raw, and
  say in notes that it was resolved from a ditto mark.
- If a person's surname is not stated (most commonly a mother in a
  baptism/burial register), infer it from another named family member in
  the same record — treat this exactly like a ditto mark: record the
  inferred surname in name_raw, flag the inference in notes.
- detail = role-specific document content that is not age/DOB (a bequest
  description, a stated occupation, a cause of death). notes = extraction-
  quality flags and provenance ONLY (a ditto resolution, an inferred
  surname, an ambiguous reading, a confidence caveat). Never mix the two.
- confidence per mention: "high" = name/role/age explicitly and clearly
  stated; "medium" = role inferred (no explicit relation column), or the
  name is unusual/hard to read; "low" = a genuine guess where an
  alternative reading is roughly as plausible.
- When a document format has no explicit relation-to-head/role column
  (e.g. 1841 Census, 1939 Register), you may infer role from position,
  marital status, or age pattern — but never force a role that contradicts
  the age pattern (e.g. do not guess "Son" for someone close in age to the
  household head's own generation just to fill the field; leave the role
  vaguer or mark it low confidence instead).
- Assume ZERO knowledge of any other document or of the family tree. Never
  write "matches X already known", "confirms Y", "same person as Z", or
  any cross-document claim in detail or notes — you have no memory of
  other documents, and identity/collision resolution is a later stage's
  job, not extraction's. Describe only what THIS page states.

### Person-Extraction Guide (doc_type_detected → role_in_record vocabulary and scoping)

{role_guidance_block}

## Empty / excluded pages

If this page has no person genuinely worth extracting — a blank or
instructions-only form page, a document unrelated to any person despite
its filename, a duplicate with no content, or content you judge sensitive
and unrelated to this family history project — set "content_flag": true
and explain why in "content_note". Leave "person_mentions" as an empty
array in that case. Otherwise set "content_flag": false and leave
"content_note" empty.

## Output Format

Your response must be a single JSON object and nothing else.

Do not include any text, explanation, or commentary before or after the JSON.
Do not wrap the JSON in markdown code fences.
Do not use ```json or ```.
Do not add a preamble sentence.
Your entire response must be valid JSON starting with {{ and ending with }}.

Use exactly this structure, choosing doc_type_detected from the vocabulary below:

Vocabulary for doc_type_detected:
{vocab_list}

{{
  "transcription": "full text of the document, preserving layout",
  "doc_type_detected": "exactly one value from the vocabulary above",
  "confidence": "high or medium or low",
  "confidence_notes": "brief note on factors affecting confidence, or No significant issues.",
  "personal_names": ["name as written", "name as written"],
  "locations": ["place as written", "place as written"],
  "uncertain_entries": [
    {{"entry": "best reading", "field": "column or field name", "reason": "why uncertain"}}
  ],
  "content_flag": false,
  "content_note": "",
  "person_mentions": [
    {{
      "name_raw": "Forename Surname",
      "role_in_record": "one value from this doc_type's guide above, or your best judgement if none is listed",
      "age_raw": "age as transcribed, or null",
      "dob_raw": "date of birth as transcribed, or null",
      "detail": "role-specific content, or empty string",
      "notes": "extraction-quality/provenance flags only, or empty string",
      "confidence": "high or medium or low"
    }}
  ]
}}"""
````

````python
def split_pdf_to_page_images(file_path: str) -> tuple:
    """
    Split a PDF into a list of JPEG image bytes using PyMuPDF.
    PyMuPDF ships MuPDF as a bundled wheel — no OS-level dependencies needed,
    so this works on Databricks serverless compute.
    Returns (list_of_page_bytes, mime_type).
    """
    import pymupdf

    doc = pymupdf.open(file_path)
    page_bytes = []
    for page in doc:
        # Render at 2x scale (~144 DPI) — good OCR quality without excessive token use
        mat = pymupdf.Matrix(2, 2)
        pix = page.get_pixmap(matrix=mat)
        page_bytes.append(pix.tobytes("jpeg"))
    doc.close()

    print(f"    PDF split into {len(page_bytes)} page image(s) via PyMuPDF")
    return page_bytes, "image/jpeg"


def extract_json(text: str) -> str:
    """
    Robustly extract a JSON object from Gemini's response text.
    Finds the outermost { ... } regardless of any surrounding text,
    preamble sentences, or markdown fences. This is more reliable than
    trying to strip specific fence patterns.
    """
    match = re.search(r'\{.*\}', text, re.DOTALL)
    if match:
        return match.group(0)
    return text  # Return as-is and let json.loads raise a clear error


def _call_gemini_once(file_bytes: bytes, mime_type: str, prompt: str) -> tuple:
    """
    Single attempt to call Gemini. Separated from retry logic so tenacity
    can wrap it cleanly without capturing non-retryable exceptions.
    Uses google-genai SDK (replaces deprecated vertexai.generative_models).
    Returns (result_dict, usage_dict).
    usage_dict keys: prompt_tokens, output_tokens, total_tokens.
    """
    image_part = types.Part.from_bytes(data=file_bytes, mime_type=mime_type)

    response = client.models.generate_content(
        model=VERTEX_MODEL,
        contents=[image_part, prompt],
        config=types.GenerateContentConfig(
            temperature=0.1,
            max_output_tokens=65536,
            thinking_config=types.ThinkingConfig(thinking_level="LOW"), # reduce Gemini 3 thinking to keep token use dow, but keep an eye on quality
            http_options=types.HttpOptions(timeout=120000),  # 2 minutes in milliseconds
        ),
    )

    # Capture token usage — always present even on safety blocks
    um = response.usage_metadata
    usage = {
        "prompt_tokens":  getattr(um, "prompt_token_count",     0) or 0,
        "output_tokens":  getattr(um, "candidates_token_count", 0) or 0,
        "thinking_tokens": getattr(um, "thoughts_token_count",    0) or 0,
        "total_tokens":   getattr(um, "total_token_count",      0) or 0,
    }

    # Detect safety block — candidates empty or content has no parts
    if not response.candidates:
        return {"_safety_blocked": True}, usage
    candidate = response.candidates[0]
    if not candidate.content or not candidate.content.parts:
        return {
            "_safety_blocked": True,
            "_finish_reason": str(getattr(candidate, "finish_reason", "unknown")),
        }, usage

    raw_text = response.text.strip()
    raw_text = extract_json(raw_text)

    try:
        return json.loads(raw_text), usage
    except json.JSONDecodeError:
        return {
            "transcription":     raw_text,
            "doc_type_detected": "Other",
            "confidence":        "low",
            "confidence_notes":  "Gemini response could not be parsed as JSON — raw text preserved.",
            "personal_names":    [],
            "locations":         [],
            "uncertain_entries": [],
            "content_flag":      False,
            "content_note":      "",
            "person_mentions":   [],
        }, usage


@retry(
    # Retry specifically on Vertex AI quota/rate-limit exceptions.
    # ResourceExhausted = gRPC quota exhaustion; TooManyRequests = HTTP 429.
    # Using the SDK exception classes directly is more reliable than string inspection.
    retry=retry_if_exception_type((
        google.api_core.exceptions.ResourceExhausted,  # 429
        google.api_core.exceptions.TooManyRequests,
        google.api_core.exceptions.DeadlineExceeded,   # 504
        google.api_core.exceptions.Cancelled,          # 499
    )),
    wait=wait_random_exponential(multiplier=1, max=60),
    stop=stop_after_attempt(5),
        before_sleep=lambda rs: print(
            f"    Rate limited (429) — attempt {rs.attempt_number} failed, retrying..."
        ),
)
def call_gemini_with_retry(file_bytes: bytes, mime_type: str, prompt: str) -> tuple:
    """
    Call Gemini with tenacity exponential backoff on 429 rate limit errors.

    Uses wait_random_exponential(multiplier=1, max=60) as recommended by Google:
    waits a random duration up to 2^attempt seconds, capped at 60 seconds.
    Jitter prevents retry storms when multiple files hit the limit simultaneously.
    Non-429 errors are raised immediately without retry.
    Returns (result_dict, usage_dict) from _call_gemini_once.
    """
    return _call_gemini_once(file_bytes, mime_type, prompt)


def convert_gif_to_jpeg(file_path: str) -> bytes:
    """
    Convert a GIF to JPEG bytes using PyMuPDF.
    Gemini nominally supports image/gif but returns 400 BadRequest on some model
    versions. Converting to JPEG is more reliable.
    """
    import pymupdf
    doc = pymupdf.open(file_path)
    page = doc[0]
    mat = pymupdf.Matrix(2, 2)
    pix = page.get_pixmap(matrix=mat)
    doc.close()
    return pix.tobytes("jpeg")


def parse_age_years(age_raw):
    """
    Best-effort parse of a transcribed age string into whole years.
    Returns None for ages given in months/weeks/days (e.g. "6 mo", "3 wks")
    rather than guessing a fractional year, and for anything with no digits.
    """
    if not age_raw:
        return None
    text = str(age_raw).lower()
    if re.search(r'\b(mo|month|wk|week|day)s?\b', text):
        return None
    match = re.search(r'\d+', text)
    return int(match.group(0)) if match else None


def ocr_file(file_path: str, metadata: dict) -> tuple:
    """
    Run OCR + person-mention extraction on a single file.
    Returns (transcription_row_dicts, person_mention_row_dicts, status_string,
             prompt_tokens, output_tokens, thinking_tokens, total_tokens).
    - Images produce one transcription row.
    - PDFs are split into pages — one transcription row per page.
    - GIFs are converted to JPEG before sending (avoids Gemini 400 BadRequest).
    - Safety blocks return ([], [], "safety_blocked", ...).
    """
    type_info = get_mime_type(file_path)
    if type_info is None:
        raise ValueError(f"Unsupported file type: {file_path}")

    mime_type, _ = type_info
    prompt = build_prompt(metadata)
    now = datetime.now(timezone.utc)

    if mime_type == "application/pdf":
        pages_bytes, page_mime = split_pdf_to_page_images(file_path)
    elif mime_type == "image/gif":
        print(f"    Converting GIF to JPEG for Gemini compatibility")
        print(f"    Commented out GIF conversion as it seems to hang")
        # pages_bytes = [convert_gif_to_jpeg(file_path)]
        # page_mime = "image/jpeg"
    else:
        with open(file_path, "rb") as f:
            pages_bytes = [f.read()]
        page_mime = mime_type

    rows = []
    mention_rows = []
    file_prompt_tokens  = 0
    file_output_tokens  = 0
    file_thinking_tokens  = 0
    file_total_tokens   = 0

    for page_index, page_bytes in enumerate(pages_bytes, 1):
        result, usage = call_gemini_with_retry(page_bytes, page_mime, prompt)

        file_prompt_tokens  += usage["prompt_tokens"]
        file_output_tokens  += usage["output_tokens"]
        file_thinking_tokens  += usage["thinking_tokens"]
        file_total_tokens   += usage["total_tokens"]

        if result.get("_safety_blocked"):
            return [], [], "safety_blocked", file_prompt_tokens, file_output_tokens, file_thinking_tokens, file_total_tokens

        content_flag = bool(result.get("content_flag", False))
        # Flag medium and low — anything that is not high warrants a check
        row_confidence_flag = result.get("confidence", "low") != "high"

        rows.append({
            "file_id":           metadata["file_id"],
            "file_name":         metadata["file_name"],
            "folder_path":       FOLDER_PATH,
            "surname":           metadata.get("surname"),
            "forename":          metadata.get("forename"),
            "year":              metadata.get("year"),
            "doc_type":          metadata.get("doc_type"),
            "doc_type_detected": result.get("doc_type_detected", "Other"),
            "page_index":        page_index,
            "transcribed_text":  result.get("transcription", ""),
            "personal_names":    json.dumps(result.get("personal_names", [])),
            "locations":         json.dumps(result.get("locations", [])),
            "uncertain_entries": json.dumps(result.get("uncertain_entries", [])),
            "confidence":        result.get("confidence", "low"),
            "confidence_notes":  result.get("confidence_notes", ""),
            "confidence_flag":   row_confidence_flag,
            "content_flag":      content_flag,
            "content_note":      result.get("content_note", "") or "",
            # PENDING whenever Gemini itself raised a concern (empty/excluded
            # content OR sub-high transcription confidence) — one unified
            # review queue; Ed sets COMPLETE after checking. NULL = no review
            # ever needed. See ocr_transcriptions.review_status comment.
            "review_status":     "PENDING" if (content_flag or row_confidence_flag) else None,
            "reviewed_at":       None,
            "review_notes":      None,
            "last_matched_at":   None,
            "gedcom_id":         None,
            "model_used":        VERTEX_MODEL,
            "processed_at":      now,
        })

        for person_index, mention in enumerate(result.get("person_mentions", []) or []):
            age_raw = mention.get("age_raw")
            mention_rows.append({
                "file_id":          metadata["file_id"],
                "page_index":       page_index,
                "person_index":     person_index,
                "name_raw":         mention.get("name_raw"),
                "role_in_record":   mention.get("role_in_record"),
                "age_raw":          age_raw,
                "age_years":        parse_age_years(age_raw),
                "dob_raw":          mention.get("dob_raw"),
                "detail":           mention.get("detail") or "",
                "notes":            mention.get("notes") or "",
                "confidence":       mention.get("confidence", "medium"),
                "model_used":       VERTEX_MODEL,
                "extracted_at":     now,
                "corrected_at":     None,
                "correction_notes": None,
            })

    return rows, mention_rows, "success", file_prompt_tokens, file_output_tokens, file_thinking_tokens, file_total_tokens
````

````python
transcription_rows = []
log_rows = []

TRANSCRIPTION_SCHEMA = StructType([
    StructField("file_id",           StringType(),    True),
    StructField("file_name",         StringType(),    True),
    StructField("folder_path",       StringType(),    True),
    StructField("surname",           StringType(),    True),
    StructField("forename",          StringType(),    True),
    StructField("year",              StringType(),    True),
    StructField("doc_type",          StringType(),    True),
    StructField("doc_type_detected", StringType(),    True),
    StructField("page_index",        IntegerType(),   True),
    StructField("transcribed_text",  StringType(),    True),
    StructField("personal_names",    StringType(),    True),
    StructField("locations",         StringType(),    True),
    StructField("uncertain_entries", StringType(),    True),
    StructField("confidence",        StringType(),    True),
    StructField("confidence_notes",  StringType(),    True),
    StructField("confidence_flag",   BooleanType(),   True),
    StructField("content_flag",      BooleanType(),   True),
    StructField("content_note",      StringType(),    True),
    StructField("review_status",     StringType(),    True),
    StructField("reviewed_at",       TimestampType(), True),
    StructField("review_notes",      StringType(),    True),
    StructField("last_matched_at",   TimestampType(), True),
    StructField("gedcom_id",         StringType(),    True),
    StructField("model_used",        StringType(),    True),
    StructField("processed_at",      TimestampType(), True),
])

PERSON_MENTION_SCHEMA = StructType([
    StructField("file_id",          StringType(),    True),
    StructField("page_index",       IntegerType(),   True),
    StructField("person_index",     IntegerType(),   True),
    StructField("name_raw",         StringType(),    True),
    StructField("role_in_record",   StringType(),    True),
    StructField("age_raw",          StringType(),    True),
    StructField("age_years",        IntegerType(),   True),
    StructField("dob_raw",          StringType(),    True),
    StructField("detail",           StringType(),    True),
    StructField("notes",            StringType(),    True),
    StructField("confidence",       StringType(),    True),
    StructField("model_used",       StringType(),    True),
    StructField("extracted_at",     TimestampType(), True),
    StructField("corrected_at",     TimestampType(), True),
    StructField("correction_notes", StringType(),    True),
])

LOG_SCHEMA = StructType([
    StructField("file_id",       StringType(),    True),
    StructField("file_name",     StringType(),    True),
    StructField("folder_path",   StringType(),    True),
    StructField("status",        StringType(),    True),
    StructField("page_count",    IntegerType(),   True),
    StructField("model_used",    StringType(),    True),
    StructField("processed_at",  TimestampType(), True),
    StructField("error_message", StringType(),    True),
])

# Write results immediately after each file so that a cancelled or failed
# run does not lose work already completed. The idempotency check at the top
# of the loop (left_anti join on the log table) ensures files already logged
# as success are not reprocessed on the next run.

def write_transcription_rows(rows: list):
    """Write a list of transcription row dicts to Delta immediately."""
    if not rows:
        return
    df = spark.createDataFrame(rows, schema=TRANSCRIPTION_SCHEMA)
    df.write.format("delta").mode("append").saveAsTable(TRANSCRIPTIONS_TBL)


def write_person_mention_rows(rows: list):
    """Write a list of person-mention row dicts to Delta immediately."""
    if not rows:
        return
    df = spark.createDataFrame(rows, schema=PERSON_MENTION_SCHEMA)
    df.write.format("delta").mode("append").saveAsTable(PERSON_MENTION_TBL)


def write_log_row(row: dict):
    """Write a single processing log row dict to Delta immediately."""
    df = spark.createDataFrame([row], schema=LOG_SCHEMA)
    df.write.format("delta").mode("append").saveAsTable(PROC_LOG_TBL)


TOKEN_USAGE_SCHEMA = StructType([
    StructField("logged_at",      TimestampType(), True),
    StructField("file_id",        StringType(),    True),
    StructField("file_name",      StringType(),    True),
    StructField("model_used",     StringType(),    True),
    StructField("page_count",     IntegerType(),   True),
    StructField("prompt_tokens",  IntegerType(),   True),
    StructField("output_tokens",  IntegerType(),   True),
    StructField("thinking_tokens",  IntegerType(),   True),
    StructField("total_tokens",   IntegerType(),   True),
])


def _write_token_usage(file_id: str, file_name: str, page_count: int,
                       prompt_tokens: int, output_tokens: int, thinking_tokens: int,
                       total_tokens: int, logged_at):
    """
    Write token usage for one file to the usage log table.
    Non-fatal — a failure here never aborts the main pipeline.
    """
    try:
        df = spark.createDataFrame([{
            "logged_at":     logged_at,
            "file_id":       file_id,
            "file_name":     file_name,
            "model_used":    VERTEX_MODEL,
            "page_count":    page_count,
            "prompt_tokens": prompt_tokens,
            "output_tokens": output_tokens,
            "thinking_tokens": thinking_tokens,
            "total_tokens":  total_tokens,
        }], schema=TOKEN_USAGE_SCHEMA)
        df.write.format("delta").mode("append").saveAsTable(TOKEN_USAGE_TBL)
    except Exception as e:
        print(f"  ⚠ Token usage logging failed (non-fatal): {e}")


# Keep summary-only accumulators for the end-of-run report.
# Individual rows are no longer held in memory.
all_rows_summary = []   # lightweight dicts: file_name, page_index, confidence, confidence_flag, confidence_notes

total            = len(to_process_df)
success_count    = 0
error_count      = 0
blocked_count    = 0
skipped_count    = 0
permanent_count  = 0
mentions_total   = 0

for i, file_row in enumerate(to_process_df, 1):
    file_path = file_row.file_path
    file_name = Path(file_path).name
    file_id   = file_row.file_id
    now       = datetime.now(timezone.utc)

    # Derive the folder path for this specific file (used in log writes)
    # relative_path is e.g. "Documents/Family_BALL/BALL_John_1891_Census.jpg"
    relative_parts = file_row.relative_path.split("/")
    FOLDER_PATH = f"{VOLUME_BASE}/{relative_parts[0]}/{relative_parts[1]}" if len(relative_parts) >= 2 else VOLUME_BASE

    print(f"[{i}/{total}] {file_name}")

    # Check file type early — avoids Gemini calls for genuinely unsupported files
    type_info = get_mime_type(file_path)
    if type_info is None:
        print(f"  ⚠ Skipped — unrecognised file type")
        write_log_row({
            "file_id": file_id, "file_name": file_name, "folder_path": FOLDER_PATH,
            "status": "skipped", "page_count": 0, "model_used": VERTEX_MODEL,
            "processed_at": now,
            "error_message": "Unrecognised file type — no extension and no matching magic bytes",
        })
        skipped_count += 1
        continue

    metadata = parse_filename(file_name)
    metadata["file_id"]   = file_id
    metadata["file_name"] = file_name

    if not metadata["parse_ok"]:
        print(f"  ⚠ Filename did not parse cleanly — skipping")
        write_log_row({
            "file_id": file_id, "file_name": file_name, "folder_path": FOLDER_PATH,
            "status": "skipped", "page_count": 0, "model_used": VERTEX_MODEL,
            "processed_at": now,
            "error_message": "Unable to parse filename",
        })
        skipped_count += 1
        continue

    try:
        rows, mention_rows, status, prompt_tokens, output_tokens, thinking_tokens, total_tokens = ocr_file(file_path, metadata)

        if status == "safety_blocked":
            print(f"  ✗ Blocked by Gemini safety filters")
            write_log_row({
                "file_id": file_id, "file_name": file_name, "folder_path": FOLDER_PATH,
                "status": "safety_blocked", "page_count": 0, "model_used": VERTEX_MODEL,
                "processed_at": now,
                "error_message": "Response blocked by Gemini safety filters. Try converting to image and reprocessing.",
            })
            _write_token_usage(file_id, file_name, 0, prompt_tokens, output_tokens, thinking_tokens, total_tokens, now)
            blocked_count += 1
            continue

        # Write transcription rows, person-mention rows, and log entry to Delta
        # immediately — results are safe even if the run is cancelled after this point.
        write_transcription_rows(rows)
        write_person_mention_rows(mention_rows)
        write_log_row({
            "file_id": file_id, "file_name": file_name, "folder_path": FOLDER_PATH,
            "status": "success", "page_count": len(rows), "model_used": VERTEX_MODEL,
            "processed_at": now, "error_message": None,
        })
        _write_token_usage(file_id, file_name, len(rows), prompt_tokens, output_tokens, thinking_tokens, total_tokens, now)

        flagged    = any(r["confidence_flag"] or r["content_flag"] for r in rows)
        confidence = rows[0]["confidence"] if rows else "unknown"
        print(f"  ✓ Done  pages={len(rows)}  mentions={len(mention_rows)}  confidence={confidence}"
              + ("  ⚑ review flagged" if flagged else ""))
        # Accumulate lightweight summary for end-of-run report
        all_rows_summary.extend([
            {"file_name": r["file_name"], "page_index": r["page_index"],
             "confidence": r["confidence"], "confidence_flag": r["confidence_flag"],
             "confidence_notes": r["confidence_notes"],
             "content_flag": r["content_flag"], "content_note": r["content_note"]}
            for r in rows
        ])
        success_count  += 1
        mentions_total += len(mention_rows)

    except Exception as e:
        error_msg = f"{type(e).__name__}: {e}\n{traceback.format_exc()}"
        attempts_so_far = error_attempt_counts.get(file_id, 0) + 1
        error_attempt_counts[file_id] = attempts_so_far

        if attempts_so_far >= MAX_RETRIES:
            # Bounded retry: this file has now failed MAX_RETRIES times total
            # (across this run and prior runs) -- give up rather than let it
            # resurface as a candidate forever. Logged as its own status,
            # distinct from 'skipped' (which means "not a real document") and
            # from the retry-eligible 'error' rows still in the log for this
            # file_id -- the full attempt history stays visible either way.
            print(f"  ✗ Error (attempt {attempts_so_far}/{MAX_RETRIES}, giving up): {error_msg[:200]}")
            write_log_row({
                "file_id": file_id, "file_name": file_name, "folder_path": FOLDER_PATH,
                "status": "failed_permanent", "page_count": 0, "model_used": VERTEX_MODEL,
                "processed_at": now,
                "error_message": f"Gave up after {attempts_so_far} attempts. Last error: {error_msg[:1800]}",
            })
            permanent_count += 1
        else:
            print(f"  ✗ Error (attempt {attempts_so_far}/{MAX_RETRIES}, will retry next run): {error_msg[:200]}")
            write_log_row({
                "file_id": file_id, "file_name": file_name, "folder_path": FOLDER_PATH,
                "status": "error", "page_count": 0, "model_used": VERTEX_MODEL,
                "processed_at": now, "error_message": error_msg[:2000],
            })
            error_count += 1

    # Pace requests to stay within Vertex AI RPM quota.
    # Placed at end of loop body so it runs after every file (success, error,
    # or safety_blocked). Skipped files use 'continue' above so don't reach here,
    # but their API cost is zero so the omission is fine.
    if i < total:
        time.sleep(INTER_REQUEST_DELAY)

print(f"\n{'='*50}")
print(f"Run complete: {success_count} succeeded, {error_count} errors (will retry), "
      f"{permanent_count} gave up (max retries reached), "
      f"{blocked_count} safety blocked, {skipped_count} skipped  ({total} candidates)")

# Print token usage totals for this run
try:
    # Build an IN-list of file_ids processed this run for the usage query
    run_file_ids = ", ".join(f"'{r.file_id}'" for r in to_process_df) if to_process_df else "''"
    usage_summary = spark.sql(f"""
        SELECT
            SUM(prompt_tokens)    AS prompt_tokens,
            SUM(output_tokens)    AS output_tokens,
            SUM(thinking_tokens)  AS thinking_tokens,
            SUM(total_tokens)     AS total_tokens,
            COUNT(*)              AS files_logged
        FROM {TOKEN_USAGE_TBL}
        WHERE file_id IN ({run_file_ids})
    """).collect()[0]
    print(f"\nToken usage (this run):")
    print(f"  Prompt tokens   : {usage_summary.prompt_tokens:,}")
    print(f"  Output tokens   : {usage_summary.output_tokens:,}")
    print(f"  Thinking tokens : {usage_summary.thinking_tokens:,}")
    print(f"  Total tokens    : {usage_summary.total_tokens:,}")
except Exception as e:
    print(f"\n⚠ Could not retrieve token usage summary: {e}")
````
