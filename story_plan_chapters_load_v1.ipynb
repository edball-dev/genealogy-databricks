# Story Plan Chapters — Load v1

**Run order:** standalone — no dependency on other notebooks. Safe to run any time.

**Purpose:** Creates `genealogy.silver_story_plan_chapters`, a one-time load mapping
each of the 60 chapters in the Sixty Stories Master Plan (Google Doc, v4) to the
`person_gedcom_id`(s) it covers. This links the plan's chapter-level tracking
(WRITTEN / DRAFT / CORE / FLEX / INTERLUDE) to person-level records, so the app's
branch tabs can show "remaining chapters for this branch" and cross-check
data-driven story candidates against what's already planned.

**Latest changes:** Initial load — all 60 chapters mapped and confirmed by Ed
(session: branch tabs planning). Chapter 46 (the Glendonwyn–Douglas–Stewart
medieval line) is anchored at the top and bottom of the line rather than mapped
generation-by-generation, by design — that chapter spans ~15 generations.

**Design choice:** table not view — this is a manually curated, infrequently
updated mapping (re-run only when the plan doc changes), not something that
should reflect "latest upstream data" automatically.


```python
%sql
-- Table not view: this is a manually curated, infrequently-updated mapping
-- (re-run only when the Sixty Stories plan doc changes), not live upstream data.
CREATE OR REPLACE TABLE genealogy.silver_story_plan_chapters (
  chapter_number     INT,
  part_name          STRING,   -- branch name (Ball/Mattock/McLean/Cuthbertson/Easter/Clifford/Pearson/Palmer), NULL for opening/closing/interludes
  title              STRING,
  status             STRING,   -- written | draft | core | flex | interlude
  person_gedcom_ids  ARRAY<STRING>,
  flags              STRING,   -- '‡' sensitive material, '⚑' research gate, both, or NULL
  doc_version        STRING,
  updated_at         TIMESTAMP
) USING DELTA;
```
